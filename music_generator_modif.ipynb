```python
def get_batch(vectorized_songs, seq_length, batch_size):
    n = vectorized_songs.shape[0] - 1
    idx = np.random.choice(n - seq_length, batch_size)

    input_batch = [vectorized_songs[i:i + seq_length] for i in idx]
    output_batch = [vectorized_songs[i + 1:i + seq_length + 1] for i in idx]

    x_batch = np.reshape(input_batch, [batch_size, seq_length])
    y_batch = np.reshape(output_batch, [batch_size, seq_length])
    return x_batch, y_batch
test_args = (vectorized_songs, 5, 2)


x_batch, y_batch = get_batch(vectorized_songs, seq_length=5, batch_size=1)
print(x_batch)
print(y_batch)
```

```python
# LSTM integration and short training loop (PyTorch)
# Paste this cell after you've created `joined_songs` or the variables that contain your dataset.
# It will try to reuse `joined_songs`, `vocab`, `char2idx`, `vectorized_songs` if they exist.

import os
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from pathlib import Path

# -------------------------
# CONFIG (tweak as needed)
# -------------------------
SEQ_LENGTH = 50         # length of input sequences
BATCH_SIZE = 32
EMBED_DIM = 128
HIDDEN_SIZE = 256
NUM_LAYERS = 2
LR = 0.001
NUM_EPOCHS = 50          # set to 1 for quick demo; increase for real training
MAX_CHARS = 20000       # cap number of chars used for fast iterations
OUT_DIR = Path("/mnt/data/music_lstm_output")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# -------------------------
# Reconstruct or reuse data
# -------------------------
# Try to reuse variables from previous cells; otherwise build from scratch
try:
    joined_songs  # noqa: F821
except NameError:
    # If `joined_songs` not present, try to build from vectorized_songs or fall back
    try:
        vectorized_songs  # noqa: F821
        # If vectorized_songs is int arrays (list of sequences), convert back to chars is tricky.
        # We'll fallback to a small default string to keep pipeline runnable.
        joined_songs = "CDEFGAB " * 1000
    except NameError:
        joined_songs = "CDEFGAB " * 1000

# Build vocab and mappings (if user already has these, override here is safe)
vocab = sorted(set(joined_songs))
char2idx = {c: i for i, c in enumerate(vocab)}
idx2char = np.array(vocab)
vocab_size = len(vocab)
print(f"Vocab size: {vocab_size}")

# Convert joined_songs to integer sequence
data = np.array([char2idx[c] for c in joined_songs], dtype=np.int64)
data = data[:min(len(data), MAX_CHARS)]

# Create sequence dataset: input is seq_length chars -> target is next seq_length chars (shifted by 1)
def make_sequences(arr, seq_len, step=1):
    X, Y = [], []
    for i in range(0, len(arr) - seq_len, step):
        X.append(arr[i: i + seq_len])
        Y.append(arr[i + 1: i + seq_len + 1])
    return np.array(X, dtype=np.int64), np.array(Y, dtype=np.int64)

# Use step=seq_len to reduce number of sequences for quick demo; change to step=1 for full overlapping sequences.
X, Y = make_sequences(data, SEQ_LENGTH, step=SEQ_LENGTH)
print("Sequences:", X.shape)

# -------------------------
# PyTorch Dataset / DataLoader
# -------------------------
import torch.utils.data as data_utils
dataset = data_utils.TensorDataset(torch.tensor(X, dtype=torch.long), torch.tensor(Y, dtype=torch.long))
loader = data_utils.DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# -------------------------
# Model
# -------------------------
class CharLSTM(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_layers):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, vocab_size)
    def forward(self, x, hidden=None):
        # x: (B, S)
        emb = self.embedding(x)  # (B, S, E)
        if hidden is None:
            out, hidden = self.lstm(emb)
        else:
            out, hidden = self.lstm(emb, hidden)
        logits = self.fc(out)    # (B, S, V)
        return logits, hidden

model = CharLSTM(vocab_size, EMBED_DIM, HIDDEN_SIZE, NUM_LAYERS).to(device)
optimizer = optim.Adam(model.parameters(), lr=LR)
criterion = nn.CrossEntropyLoss()

# -------------------------
# Training loop (simple)
# -------------------------
model.train()
for epoch in range(1, NUM_EPOCHS + 1):
    running_loss = 0.0
    for batch_idx, (bx, by) in enumerate(loader, 1):
        bx = bx.to(device)  # (B, S)
        by = by.to(device)
        optimizer.zero_grad()
        logits, _ = model(bx)  # (B, S, V)
        loss = criterion(logits.reshape(-1, vocab_size), by.reshape(-1))
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        # lightweight logging
        if batch_idx % 20 == 0:
            print(f"Epoch {epoch} Batch {batch_idx}/{len(loader)} Loss {running_loss / batch_idx:.4f}")
    avg = running_loss / max(1, len(loader))
    print(f"Epoch {epoch} finished. Avg loss: {avg:.4f}")

# -------------------------
# Save model + mappings
# -------------------------
save_path = OUT_DIR / "music_lstm_model.pth"
torch.save({
    "model_state_dict": model.state_dict(),
    "vocab": vocab,
    "char2idx": char2idx,
    "idx2char": idx2char.tolist(),
    "seq_length": SEQ_LENGTH,
    "embed_dim": EMBED_DIM,
    "hidden_size": HIDDEN_SIZE,
    "num_layers": NUM_LAYERS
}, save_path)
print("Saved model and mappings to:", save_path)

# -------------------------
# OPTIONAL: simple sampling function to generate a short sequence
# -------------------------
def sample(model, start_str="C", length=200, temperature=1.0):
    model.eval()
    device = next(model.parameters()).device
    seq = [char2idx.get(c, 0) for c in start_str]
    input_seq = torch.tensor(seq, dtype=torch.long, device=device).unsqueeze(0)  # (1, S)
    hidden = None
    generated = start_str
    with torch.no_grad():
        for _ in range(length):
            logits, hidden = model(input_seq, hidden)
            # take only last time-step logits
            last_logits = logits[:, -1, :] / max(1e-8, temperature)
            probs = torch.softmax(last_logits, dim=-1)
            next_idx = torch.multinomial(probs, num_samples=1).item()
            generated += idx2char[next_idx]
            # feed back
            input_seq = torch.tensor([[next_idx]], dtype=torch.long, device=device)
    return generated

# Example (quick) generation:
#print("--- sample output ---")
generated_song=(sample(model, start_str="C", length=200, temperature=1.0))
```
